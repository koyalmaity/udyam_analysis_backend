```python
def get_category(row):
    basic = str(row["basic_category"]).lower()
    
    taxonomy = row["taxonomy"]
    taxonomy_text = str(taxonomy).lower()

    # Finance
    if any(x in basic or x in taxonomy_text for x in [
        "bank", "finance", "loan", "credit", "atm", "insurance"
    ]):
        return "finance"

    # Healthcare
    if any(x in basic or x in taxonomy_text for x in [
        "pharmacy", "hospital", "clinic", "medical",
        "health", "doctor", "dentist", "wellness"
    ]):
        return "healthcare"

    # Food & Grocery
    if any(x in basic or x in taxonomy_text for x in [
        "restaurant", "food", "grocery", "supermarket",
        "bakery", "cafe", "fruit", "vegetable"
    ]):
        return "food_and_grocery"

    # Agriculture
    if any(x in basic or x in taxonomy_text for x in [
        "agricultural", "farm", "tractor", "fertilizer",
        "seed", "agro", "agriculture"
    ]):
        return "agriculture"

    # Education
    if any(x in basic or x in taxonomy_text for x in [
        "school", "college", "education", "learning",
        "training", "coaching"
    ]):
        return "education"

    # Automotive
    if any(x in basic or x in taxonomy_text for x in [
        "auto", "automotive", "vehicle", "car", "motorcycle",
        "repair", "tyre"
    ]):
        return "automotive"

    # Retail
    if any(x in basic or x in taxonomy_text for x in [
        "shopping", "retail", "clothing", "fashion",
        "store", "shop", "market"
    ]):
        return "retail"

    # Professional services
    if any(x in basic or x in taxonomy_text for x in [
        "professional", "office", "accounting",
        "legal", "consulting", "advertising"
    ]):
        return "professional_services"

    # Transport
    if any(x in basic or x in taxonomy_text for x in [
        "transport", "train_station", "bus",
        "railway", "taxi"
    ]):
        return "transport"

    return "other"
```

```python
def map_business_group(row):
    basic = str(row["basic_category"]).lower()
    primary = str(row["taxonomy_primary"]).lower()

    text = basic + " " + primary

    # Finance
    if any(x in text for x in [
        "bank", "finance", "credit", "loan", "atm", "insurance"
    ]):
        return "finance"

    # Healthcare
    if any(x in text for x in [
        "health", "medical", "clinic", "pharmacy",
        "medicine", "doctor", "hospital", "naturopathic",
        "outpatient", "counseling"
    ]):
        return "healthcare"

    # Agriculture
    if any(x in text for x in [
        "agriculture", "agricultural", "farm",
        "dairy", "fertilizer", "seed", "krushi"
    ]):
        return "agriculture"

    # Food & Grocery
    if any(x in text for x in [
        "grocery", "food", "restaurant", "cafe",
        "bakery", "sweet", "food_and_beverage"
    ]):
        return "food_and_grocery"

    # Automotive
    if any(x in text for x in [
        "auto", "automotive", "vehicle", "motorcycle",
        "car", "tyre", "repair"
    ]):
        return "automotive"

    # Retail
    if any(x in text for x in [
        "shopping", "store", "shop", "retail",
        "clothing", "fashion", "jewelry", "shoe",
        "hardware", "furniture", "electronics",
        "mobile", "bookstore"
    ]):
        return "retail"

    # Education
    if any(x in text for x in [
        "education", "school", "college",
        "educational", "training"
    ]):
        return "education"

    # Hospitality
    if any(x in text for x in [
        "hotel", "lodging", "lodge", "hospitality"
    ]):
        return "hospitality"

    # Services
    if any(x in text for x in [
        "service", "printing", "real_estate",
        "advertising", "marketing", "caterer",
        "party_and_event"
    ]):
        return "services"

    # Transport
    if any(x in text for x in [
        "transport", "bus", "train", "station",
        "gas_station", "fueling"
    ]):
        return "transport"

    # Entertainment / sports
    if any(x in text for x in [
        "movie", "theater", "gym", "sports",
        "sporting"
    ]):
        return "entertainment_sports"

    # Government / community
    if any(x in text for x in [
        "police", "government", "community",
        "social"
    ]):
        return "government_community"

    # Religious
    if any(x in text for x in [
        "temple", "hindu_place_of_worship",
        "religious", "mosque", "church"
    ]):
        return "religious"

    return "other"
```

```python
from sqlalchemy import text

warud_lat = 21.4710
warud_lon = 78.2690
radius_m = 1000

with engine.connect() as connection:
    result = connection.execute(text("""
        SELECT
            business_name,
            business_group,
            ROUND(
                ST_Distance(
                    geom::geography,
                    ST_SetSRID(
                        ST_MakePoint(:lon, :lat),
                        4326
                    )::geography
                )::numeric,
                2
            ) AS distance_m
        FROM competitors
        WHERE ST_DWithin(
            geom::geography,
            ST_SetSRID(
                ST_MakePoint(:lon, :lat),
                4326
            )::geography,
            :radius_m
        )
        ORDER BY distance_m;
    """), {
        "lat": warud_lat,
        "lon": warud_lon,
        "radius_m": radius_m
    })

    rows = result.fetchall()

print(f"Competitors within {radius_m/1000:.0f} km:", len(rows))

for row in rows[:10]:
    print(row)
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("""
        SELECT
            business_group,
            COUNT(*) FILTER (
                WHERE ST_DWithin(
                    geom::geography,
                    ST_SetSRID(
                        ST_MakePoint(:lon, :lat),
                        4326
                    )::geography,
                    1000
                )
            ) AS within_1km,

            COUNT(*) FILTER (
                WHERE ST_DWithin(
                    geom::geography,
                    ST_SetSRID(
                        ST_MakePoint(:lon, :lat),
                        4326
                    )::geography,
                    3000
                )
            ) AS within_3km,

            COUNT(*) FILTER (
                WHERE ST_DWithin(
                    geom::geography,
                    ST_SetSRID(
                        ST_MakePoint(:lon, :lat),
                        4326
                    )::geography,
                    5000
                )
            ) AS within_5km

        FROM competitors
        GROUP BY business_group
        ORDER BY within_1km DESC;
    """), {
        "lat": warud_lat,
        "lon": warud_lon
    })

    rows = result.fetchall()

print("Category-wise competitor density:\n")

for row in rows:
    print(
        f"{row[0]:25} "
        f"1 km: {row[1]:3} | "
        f"3 km: {row[2]:3} | "
        f"5 km: {row[3]:3}"
    )
```

```python
import numpy as np
import geopandas as gpd
from shapely.geometry import box

# Get competitor coordinates
with engine.connect() as connection:
    df_points = pd.read_sql(
        text("""
            SELECT
                business_name,
                business_group,
                lat,
                lon
            FROM competitors
            WHERE business_group = :category
        """),
        connection,
        params={"category": selected_category}
    )

# Convert to GeoDataFrame
gdf_points = gpd.GeoDataFrame(
    df_points,
    geometry=gpd.points_from_xy(
        df_points["lon"],
        df_points["lat"]
    ),
    crs="EPSG:4326"
)

print("Competitors:", len(gdf_points))
```

```python
# Get bounding box of competitor area
minx, miny, maxx, maxy = gdf_points.total_bounds

# Grid cell size in degrees
cell_size = 0.005

grid = []

x = minx
while x < maxx:
    y = miny

    while y < maxy:
        grid.append(
            box(
                x,
                y,
                x + cell_size,
                y + cell_size
            )
        )
        y += cell_size

    x += cell_size

grid_gdf = gpd.GeoDataFrame(
    {"geometry": grid},
    crs="EPSG:4326"
)

print("Grid cells:", len(grid_gdf))
```

```python
joined = gpd.sjoin(
    gdf_points,
    grid_gdf,
    predicate="within",
    how="left"
)

density = (
    joined
    .groupby("index_right")
    .size()
    .rename("competitor_count")
)

grid_gdf["competitor_count"] = (
    density
    .reindex(grid_gdf.index)
    .fillna(0)
)

print(grid_gdf[["competitor_count"]].describe())
```
